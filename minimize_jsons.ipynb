```python
import json
import glob

from help_code import load_json, save_json
```

```python
def round_value(value):
    if isinstance(value, float):
        abs_val = abs(value)
        if abs_val == 0:
            return 0
        elif abs_val >= 1000:
            return round(value)
        elif abs_val >= 10:
            return round(value, 1)
        elif abs_val >= 1:
            return round(value, 2)
        elif abs_val >= 0.01:
            return round(value, 3)
        else:
            return float(f"{value:.2e}")
    return value

def minimize_json(data):
    if isinstance(data, dict):
        return {k: minimize_json(v) for k, v in data.items()}
    elif isinstance(data, list):
        return [minimize_json(v) for v in data]
    else:
        return round_value(data)

def calc_json_size(data):
    json_str = json.dumps(data, separators=(',', ':'))
    return len(json_str)
```

```python
files = glob.glob('json/*.json')

for f in files:
    if 'min' not in f:
        json1 = load_json(f)
        json2 = minimize_json(json1)

        print(f'{f}: orig JSON={calc_json_size(json1)}, minimized JSON={calc_json_size(json2)}')

        name_out = f.replace('.json', '_min.json')
        save_json(json2, name_out, minimal=True)
```
